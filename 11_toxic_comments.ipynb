```python
def load_csv(file_path, file_url=None, date_columns=None, index_column=None):
    """
    Загружает данные из локального CSV файла или по URL.

    Функция проверяет, существует ли файл по указанному пути. 
    Если файл существует, данные загружаются из него. 
    В противном случае данные загружаются по указанному URL, если он предоставлен.

    Параметры:
    file_path (str): Путь к локальному файлу CSV.
    file_url (str, optional): URL для загрузки данных, если файл не найден.
    date_columns (list, optional): Список имен столбцов для преобразования в даты.
    index_column (str, int, optional): Имя или индекс столбца, который будет использоваться в качестве индекса.

    Возвращает:
    DataFrame или None: Загруженные данные в формате DataFrame, или None в случае ошибки.
    """
    try:
        # Проверяем, существует ли файл по указанному пути
        if os.path.exists(file_path):
            return pd.read_csv(file_path, parse_dates=date_columns, index_col=index_column)
        elif file_url:
            # Загружаем данные из URL, если файл не найден
            return pd.read_csv(file_url, parse_dates=date_columns, index_col=index_column)
        else:
            print("File not found and no URL provided.")
            return None
    except Exception as e:
        # В случае ошибки выводим сообщение
        print("Something went wrong:", e)
        return None
```

```python
# Создаем функцию для вывода общей информации о датасете
def df_info(data):
    """
    Эта функция выводит общую информацию о переданном датасете, включая:
    1. Первые 5 строк для предварительного просмотра.
    2. Общую информацию о типах данных и количестве ненулевых значений.
    3. Статистику по колонкам, включая описательные статистики.
    4. Долю пропусков в каждом столбце.
    5. Количество полных дубликатов в датасете.
    
    Параметры:
    data : pd.DataFrame
        Датасет, для которого нужно вывести информацию.
    """
    
    # 1. Выводим первые 5 строк датасета для предварительного просмотра
    print(f'1. Первые 5 строк датасета:')
    display(data.head())
          
    # 2. Выводим общую информацию о датасете
    print('\n2. Общая информация о датасете:')
    print(data.info())
    
    # 3. Выводим статистику по колонкам
    print('\n3. Статистика по колонкам:')
    display(data.describe(include='all', datetime_is_numeric=True))
    
    # 4. Вычисляем и отображаем долю пропусков в колонках
    print('\n4. Доля пропусков в колонках:')
    missing_data = pd.DataFrame(round(data.isna().mean() * 100, 4).sort_values())
    display(missing_data.style.background_gradient('coolwarm'))
    
    # 5. Подсчитываем и выводим количество полных дубликатов
    print('\n5. Количество полных дубликатов:', data.duplicated().sum())
```

Поскольку корпус наших текстов составлен на латинице, то для лемматизации поищем библиотеки которые работают с такими текстами. В открытом доступе удалось найти информацию о следующих библиотеках:

*Источник: https://www.geeksforgeeks.org/python-lemmatization-approaches-with-examples/*

WordNet и WordNet (with POS tag) отличаются в том, как они выполняют лемматизацию, то есть приведение слова к базовой форме. Основное различие заключается в учёте частей речи (POS — Part of Speech) при лемматизации.

1. **WordNet (без POS tag)**
Когда мы используем лемматизатор WordNetLemmatizer без указания части речи (POS), он по умолчанию считает, что все слова являются существительными (noun). Это может привести к неправильной лемматизации других частей речи (например, глаголов, прилагательных).
2. **WordNet (с POS tag)**
Когда мы указываем часть речи (POS), WordNetLemmatizer использует её, чтобы точно определить базовую форму слова. Это позволяет лемматизатору корректно обрабатывать глаголы, прилагательные и наречия, а не только существительные.

Для лемматизации попробуем использовать `WordNet (with POS tag)`

```python
# Определяем функцию лемматизации текста
def lemmatize_text(sentence):
    # продублируем импорты внутри функции, чтобы можно было далее запараллелить вычисления 
    import nltk  
    from nltk.stem import WordNetLemmatizer
    from nltk.corpus import wordnet

    lemmatizer = WordNetLemmatizer()

    # Определяем функцию для получения POS тегов
    # функция определена внутри другой функции, чтобы она работала в паралельных вычислениях
    def pos_tagger(nltk_tag):
        if nltk_tag.startswith('J'):
            return wordnet.ADJ
        elif nltk_tag.startswith('V'):
            return wordnet.VERB
        elif nltk_tag.startswith('N'):
            return wordnet.NOUN
        elif nltk_tag.startswith('R'):
            return wordnet.ADV
        else:          
            return None

    # Токенизируем предложение и получаем POS теги
    pos_tagged = nltk.pos_tag(nltk.word_tokenize(sentence))
    
    # Применяем свою функцию для упрощения тегов
    wordnet_tagged = list(map(lambda x: (x[0], pos_tagger(x[1])), pos_tagged))
    
    lemmatized_sentence = []
    for word, tag in wordnet_tagged:
        if tag is None:
            # Если тег отсутствует, добавляем слово как есть
            lemmatized_sentence.append(word)
        else:        
            # Иначе используем тег для лемматизации слова
            lemmatized_sentence.append(lemmatizer.lemmatize(word, tag))
    
    # Объединяем лемматизированные слова в одно предложение
    return " ".join(lemmatized_sentence)



# Пример использования функции
sentence = 'the cat is sitting with the bats on the striped mat under many badly flying geese'
lemmatized_result = lemmatize_text(sentence)

print(lemmatized_result)
```

```python
# т.к. нащ датасет достаточно большой, оценим примерное время лемматизации всех текстов

sample_size = 1000  # Размер выборки
num_records = len(data)  # Общее количество записей в датасете

# Выбираем случайные записи
sample_data = data.sample(n=sample_size, random_state=RANDOM_STATE)

# Функция для оценки времени работы лемматизации
def evaluate_lemmatization_performance(sample):
    start_time = time()
    sample['lemmatized_text'] = sample['cleaned_text'].apply(lemmatize_text)
    end_time = time()
    return end_time - start_time

# Оценка времени работы
elapsed_time = evaluate_lemmatization_performance(sample_data)

# Рассчитываем приблизительное время лемматизации всего датасета
approx_total_time = (elapsed_time / sample_size) * num_records

# Вывод результата
print(f"Время лемматизации {sample_size} записей: {elapsed_time:.2f} секунд")
print(f"Приблизительное время лемматизации всего датасета: {approx_total_time:.2f} секунд")
```

## Обучение на n-gramm

Ранее мы создали подготовленные выборки X_train и X_test, содержащие лемматизированные корпусы текстов. Однако для применения моделей машинного обучения из каждого такого текста необходимо извлечь или создать признаки, на которых такие модели будут обучаться и искать закономерности.

Для создания признаков из текстов применим следующий подход:

1. **Векторизация текста:**
Применим методы векторизации, такие как n-граммы. Это позволит преобразовать текст в числовую форму, сохраняя контекст и последовательность слов. Будем использовать такие функции как CountVectorizer или TfidfVectorizer из sklearn.
2. **Количество слов:**
Добавим новый признак, который будет представлять количество слов в каждом тексте. Это можно легко сделать с помощью функции подсчета слов.
3. **Оценка тональности:**
Используем SentimentIntensityAnalyzer из библиотеки nltk для оценки тональности текста. Этот анализатор предоставит такие метрики, как положительная, отрицательная и нейтральная оценки, а также общее значение тональности.

Векторизаторы должны обучаться только на обучающей выборке и чтобы при кросс-валидации не происходила утечка данных. Поэтому мы  реализуем их внутри пайплайна.

Количество слов и оценка тональности рассчитываются для каждого текста отдельно и являются независимыми от других записей. Чтобы не дублировать вычисления при подборе гиперпараметров лучше эти признаки создать и добавим в выборки отдельно до момента обучения моделей.

```python
# переведем выборки с независимыми признаками в формат df
X_train_df = pd.DataFrame({'text' : X_train})
X_test_df = pd.DataFrame({'text' : X_test})
```

```python
# Функция для подсчета количества слов
def word_count(texts):
    return np.array([len(text.split()) for text in texts])

# Функция для сентиментального анализа
def sentiment_analysis(texts):
    sia = SentimentIntensityAnalyzer()
    return np.array([sia.polarity_scores(text)['compound'] for text in texts])
```

```python
# сводная функция
def add_features(df):
    # Count words
    df['word_count'] = word_count(df['text'])
    # Perform sentiment analysis
    df['sentiment'] = sentiment_analysis(df['text'])
    return df
```

```python
def find_best_model(pipe, param_grid, X_train, y_train, scoring, search_type='grid', cv=3, n_jobs=-1, n_iter=10):
    # 1. Инициализация GridSearchCV или RandomizedSearchCV
    if search_type == 'grid':
        search = GridSearchCV(
            pipe,
            param_grid,
            cv=cv,
            scoring=scoring,
            n_jobs=n_jobs
        )
    elif search_type == 'random':
        search = RandomizedSearchCV(
            pipe,
            param_distributions=param_grid,
            n_iter=n_iter,
            cv=cv,
            scoring=scoring,
            n_jobs=n_jobs
        )
    else:
        raise ValueError("search_type должен быть 'grid' или 'random'.")

    # 2. Обучение модели на тренировочных данных
    search.fit(X_train, y_train)

    # 3. Сохранение результатов поиска  
    model = search.best_estimator_
    score = abs(search.best_score_)
    results = pd.DataFrame(search.cv_results_)

    # 4. Вывод информации о лучшей модели и её параметрах
    print('Лучшая модель и её параметры:')
    print(model)

    # 5. Вывод лучшей метрики на кросс-валидации
    print('Лучшая метрика на кросс-валидации для лучшей модели:', score)

    # 6. Вывод результатов поиска
    relevant_columns = ['mean_fit_time', 'mean_score_time', 'params', 'mean_test_score']
    
    # Проверка наличия 'param_models'
    if 'param_models' in results.columns:
        relevant_columns.insert(0, 'param_models')  # Добавляем в начало списка

    results = results[relevant_columns].sort_values(by='mean_test_score', ascending=False)
    
    return search, results, score
```

```python
# Определите функцию цели для оптимизации
def objective(params):
    global results_df  # Делаем DataFrame глобальным, чтобы к нему можно было обращаться

    c = params['C']
    max_iter = params['max_iter']
    class_weight = params['class_weight']
    
    # Определяем векторизатор на основе выбранного типа
    vectorizer_choice = params['vectorizer']['type']
    
    if vectorizer_choice == 'unigram':
        vectorizer = CountVectorizer(stop_words=stop_words)
    elif vectorizer_choice == 'tfidf':
        vectorizer = TfidfVectorizer(stop_words=stop_words)
    else:
        vectorizer = 'passthrough'

    # Определяем ColumnTransformer
    preprocessor = ColumnTransformer(
        transformers=[
            ('vectorizer', vectorizer, 'text')
        ],
        remainder='passthrough'
    )

    # Определяем полный пайплайн
    pipe = Pipeline([
        ('preprocessor', preprocessor),
        ('scaler', 'passthrough'),
        ('model', LogisticRegression(
            random_state=RANDOM_STATE,
            class_weight=class_weight,
            C=c,
            max_iter=max_iter
        ))
    ])

    # Оценка модели с использованием кросс-валидации
    scores = cross_val_score(pipe, X_train_df, y_train, cv=3, scoring='f1')
    
    mean_test_score = scores.mean()  # Средний тестовый скор

    # Записываем результаты в DataFrame
    results_df = results_df.append({
        'mean_fit_time': None,
        'mean_score_time': None,
        'params': params,
        'mean_test_score': mean_test_score
    }, ignore_index=True)

    # Логирование результатов
    print(f"Evaluating: C={c}, max_iter={max_iter}, vectorizer={vectorizer_choice}, class_weight={class_weight}...")
    print(f"    F1 Score={mean_test_score:.4f}")

    return -mean_test_score  # Hyperopt минимизирует, возвращаем отрицательное значение
```

## Обучение на эмбеддингах BERT

### Подготовка данных

```python
print(X_train.shape)
print(X_test.shape)
```

Наши выборки достаточно большие и модель Берт будет очень долго выполнять токенизацию текстов и создание эмбеддингов (на текущих выборках Bert создавал бы эмбеддинги около 18 часов на моем пк).

Чтобы все таки протестировать эту модель и сравнить ее с другими попробуем взять 5% обучающей выборки и 10% тестовой.

```python
# Для тренировочного набора
X_train_subset, _, y_train_subset, _ = train_test_split(
    X_train, y_train, 
    train_size=0.05, 
    random_state=RANDOM_STATE, 
    stratify=y_train
)

# Для тестового набора
X_test_subset, _, y_test_subset, _ = train_test_split(
    X_test, y_test, 
    train_size=0.1, 
    random_state=RANDOM_STATE, 
    stratify=y_test
)
```

```python
# Выводим формы выборок
print(f"Форма обучающей выборки: {len(X_train_subset)} текстов, {len(y_train_subset)} меток")
print(f"Форма тестовой выборки: {len(X_test_subset)} текстов, {len(y_test_subset)} меток")
```

```python
# Инициализация токенизатора и модели
model = transformers.AutoModel.from_pretrained('unitary/toxic-bert')
tokenizer = transformers.AutoTokenizer.from_pretrained('unitary/toxic-bert')
model.eval()  # Установка модели в режим оценки
```

```python
X_train_subset = pd.Series(X_train_subset)
X_test_subset = pd.Series(X_test_subset)
```

```python
def tokenize_function(text):
    return tokenizer.encode_plus(
        text,
        max_length=512,
        padding='max_length',
        truncation=True,  # Обрезать последовательности, превышающие max_length
        add_special_tokens=True,
        return_tensors='pt'  # Возвращает данные в формате PyTorch
    )
```

```python
# Использование tqdm для отслеживания прогресса
tqdm.pandas(desc="Tokenizing texts")

# Применяем токенизацию с tqdm
X_train_subset_toc = X_train_subset.progress_apply(tokenize_function)
X_test_subset_toc = X_test_subset.progress_apply(tokenize_function)
```

```python
X_train_subset_toc.shape
```

```python
def extract_embeddings(tokenized_texts):
     # Определение устройства (CPU или GPU)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.to(device)
    
    # Извлечение input_ids и attention_mask
    padded = torch.cat([t['input_ids'] for t in tokenized_texts]).squeeze(1).to(device)
    attention_mask = torch.cat([t['attention_mask'] for t in tokenized_texts]).squeeze(1).to(device)

    # Проверка размеров
    print(f"Общее количество образцов: {padded.shape[0]}")

    # Создание эмбеддингов по батчам
    batch_size = 256
    embeddings = []

    # Обработка данных по батчам
    for i in tqdm(range((padded.shape[0] + batch_size - 1) // batch_size), desc="Обработка батчей"):
        start = batch_size * i
        end = min(start + batch_size, padded.shape[0])  # Убедимся, что не выходим за пределы

        # Подготовка батча
        batch = padded[start:end]
        attention_mask_batch = attention_mask[start:end]

        with torch.no_grad():
            batch_embeddings = model(batch, attention_mask=attention_mask_batch)

        # Проверка, сколько эмбеддингов возвращается
        print(f"Батч {i + 1}: обработано {batch.shape[0]} образцов, получено {batch_embeddings[0][:,0,:].shape[0]} эмбеддингов")
        
        embeddings.append(batch_embeddings[0][:,0,:].cpu().numpy())  # Перемещаем на CPU перед конвертацией в numpy

    # Объединение всех эмбеддингов
    features = np.concatenate(embeddings)
    print(f"Всего сгенерировано эмбеддингов: {features.shape[0]}")
    return features
```

```python
# Получение эмбеддингов для обучающей выборки (выполняется около 45 минут)
train_embeddings = extract_embeddings(X_train_subset_toc)
```

```python
# Получение эмбеддингов для тестовой выборки (выполняется около 25 минут)
test_embeddings = extract_embeddings(X_test_subset_toc)
```

```python
# переименуем переменные для удобства
X_train_bert = train_embeddings
y_train_bert = y_train_subset

X_test_bert = test_embeddings
y_test_bert = y_test_subset
```

```python
# Выводим формы выборок
print(f"Форма обучающей выборки: {len(X_train_bert)} текстов, {len(y_train_bert)} меток")
print(f"Форма тестовой выборки: {len(X_test_bert)} текстов, {len(y_test_bert)} меток")
```

### Обучение моделей

#### LogisticRegression

```python
# Создаем DataFrame для хранения результатов
results_df_bert = pd.DataFrame(columns=['mean_fit_time', 'mean_score_time', 'params', 'mean_test_score'])
```

```python
# Определите целевую функцию
def objective_bert(params):
   
    global results_df_bert  # Делаем DataFrame глобальным, чтобы к нему можно было обращаться
    
    c = params['C']
    
    pipe = Pipeline([
        ('model', LogisticRegression(
            random_state=RANDOM_STATE, 
            C=c
        )),
    ])
    
    # Логирование результатов
    print(f"Evaluating: C={c}...")
    
    score = cross_val_score(pipe, X_train_bert, y_train_bert, cv=3, scoring='f1').mean()
    
    # Записываем результаты в DataFrame
    results_df_bert = results_df_bert.append({
        'mean_fit_time': None,
        'mean_score_time': None,
        'params': params,
        'mean_test_score': score
    }, ignore_index=True)
    

    print(f"    F1 Score={score:.4f}")
    
    return -score  # Hyperopt минимизирует, возвращаем отрицательное значение
```

```python
# Определите пространство параметров для перебора
space_bert = {
    'C': hp.quniform('C', 0.1, 30, 0.5),  # Прямое равномерное распределение от 0.1 до 30 с шагом 0.5
}
```

```python
# Запускаем оптимизацию
trials_bert = Trials()
best_bert = fmin(fn=objective_bert, space=space_bert, algo=tpe.suggest, max_evals=20, trials=trials_bert)

print("Best: {}".format(best_bert))
```

Несмотря на сокрашение размера обучающей выборки, качество модели на эмбеддингах BERT существенно выросло!
