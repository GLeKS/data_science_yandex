# Исследование рынка компьютерных игр и игровых платформ

В данном исследовании проводится анализ рынка компьютерных игр и игровых платформ для интернет-магазина "Стримчик".

**Цель исследования:** Проанализировать закономерности,  определяющие успешность игры, на основе исторических данных о продажах игр, оценках пользователей и экспертов, жанрах и платформах за 2016 год. Результаты анализа должны помочь интернет-магазину "Стримчик" определить потенциально популярные продукты и спланировать эффективные рекламные кампании.

**Описание исходных данных для анализа:**
Исходные данные для анализа представляют собой исторические данные о продажах игр, оценки пользователей и экспертов, жанры и платформы (например, Xbox или PlayStation) за 2016 год, взятые из открытых источников.

Данные для анализа хранятся в файле `games.csv`. Краткое описание содержания исходного файла представлено ниже:

- *Name* — название игры
- *Platform* — платформа
- *Year_of_Release* — год выпуска
- *Genre* — жанр игры
- *NA_sales —* продажи в Северной Америке (миллионы проданных копий)
- *EU_sales* — продажи в Европе (миллионы проданных копий)
- *JP_sales* — продажи в Японии (миллионы проданных копий)
- *Other_sales —* продажи в других странах (миллионы проданных копий)
- *Critic_Score* — оценка критиков (максимум 100)
- *User_Score* — оценка пользователей (максимум 10)
- *Rating* — рейтинг от организации *ESRB* (англ. *Entertainment Software Rating Board).* Эта ассоциация определяет рейтинг компьютерных игр и присваивает им подходящую возрастную категорию.

Данные за 2016 год могут быть неполными.

Но стоит проверить, есть ли игры, релизы по которым на разных платформах состоялись в разные годы. Также оценим количество таких игр.

```python
data.query('year_of_release.isna()')['platform'].unique()
```

```python
# найдем игры, у которых было несколько релизов в разные годы
year_not_unique = list( # названия игр сохраним в список
    (data
        .groupby(['name', 'year_of_release']) # сгруппируем игры по name и year_of_release
        .size() # посчитаем их количество в каждой группе
        .reset_index(name='count') # перезапишем мультииндекс таблицы в виде обычных столбцов 
        .loc[data # из получившейся таблицы выберем только те игры, у которых было несколько релизов в разные годы, 
             #для этого применим маску с дубликатами по колонке "name" методом loc
             # duplicated нужно применить к той же таблице, что получили выше, поэтому еще раз пропишем параметры группировки
            .groupby(['name', 'year_of_release']) # 
            .size()
            .reset_index(name='count')
            .duplicated(subset='name') # отметим дубликаты в колонке 'name' значениями True
            ]
    )['name'] # нас интересует только названия игр, у которых релиз был в разные годы
    # поэтому выберем столбец 'name' для записи названий в список year_not_unique
)
```

```python
# проверим, что данные отобрались верно
data.query('name in @year_not_unique').sort_values(by=['name', 'year_of_release'])
```

```python
# посмотрим у каких игр было больше всего релизов
data.query('name in @year_not_unique').sort_values(by=['name', 'year_of_release'])[['name']].value_counts()
```

**Гипотеза.** Проверим гиппотезу о равенстве средних значений продаж в строках с weekly и без.

- H0 (нулевая гипотеза) = средние продажи игр с припиской weekly и без равны.
- H1 (альтернативная гипотеза) = средние продажи игр с припиской weekly и без различаются.

Продажи игр с weekly и без weekly не зависят друг от друга, поэтому для проверки гипотез можем применить двухвыборочный t-тест для двух независимых выборок.

```python
# создадим соответствующие выборки
weekly_sales = data.query('name.str.contains("weekly", case=False )')[['na_sales', 'eu_sales', 'jp_sales', 'other_sales']]
weekly_sales['total_sales'] = weekly_sales['na_sales'] + weekly_sales['eu_sales'] + weekly_sales['jp_sales'] + weekly_sales['other_sales']

not_weekly_sales = data.query('~name.str.contains("weekly", case=False )')[['na_sales', 'eu_sales', 'jp_sales', 'other_sales']]
not_weekly_sales['total_sales'] = not_weekly_sales['na_sales'] + not_weekly_sales['eu_sales'] + not_weekly_sales['jp_sales'] + not_weekly_sales['other_sales']
```

```python
print(f'Средние продажи игра с weekly:', round(weekly_sales['total_sales'].mean(), 2))
print(f'Средние продажи игра с без weekly:', round(not_weekly_sales['total_sales'].mean(), 2))
```

```python
#Проверим действительно ли выборки независимые
# Вычисление коэффициента корреляции с помощью метода corr()

print('Коэффициент корреляции: {:.2f}'.format(weekly_sales['total_sales'].corr(not_weekly_sales['total_sales'])))
```

```python
# выбранный уровень статистической значимости
alpha = 0.01
print(f"Уровень статистической значимости: {alpha * 100}%")

# проведем двусторонний ttest
results = st.ttest_ind(weekly_sales['total_sales'], not_weekly_sales['total_sales'])

#выведем полученное значение p-value
print('p-value:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу')
```

<u>Результат проверки:</u> На имеющихся данных, при уровне значимости 1% (уровне доверия 99%) нет оснований отвергнуть нулевую гипотезу (H0) о равенстве средних продаж игр с weekly и без weekly.

```python
# определим функцию которая будет убирать из названия значения в скобках, если в них указано sales.
# так мы сможем далее отфитровать названия игр, имеющих неявные дубликаты в таблице.
def clear_names(row):
    if 'sales' in row.lower():
        return row.split('(', 1)[0].strip() # разделим строку по первой скобке и возьмем первую ее часть. 
    # также удалим лишние пробелы в начале и конце строки
    else:
        return row
```

```python
# определим функцию, которая будет суммировать продажи по дублирующимся строкам (по играм с одинаковым названием, жанром, платформой, годом выпуска)
def aggregate(data):
    data_filtered = data.groupby(['name', 'platform', 'year_of_release', 'genre'], dropna=False).agg({
        'na_sales': 'sum',
        'eu_sales': 'sum',
        'jp_sales': 'sum',
        'other_sales': 'sum',
        'critic_score': 'first',
        'user_score': 'first',
        'rating': lambda x: next(filter(lambda r: r != 'Unknown', x), 'Unknown')
# лямбда-функция принимает аргумент x, который представляет значения столбца 'rating' для каждой группы.
# используем функцию filter для фильтрации значений x, и затем next возвращает первое непустое значение или 'Unknown', если такого значения нет.
        
    }).fillna(np.nan).sort_values(by='name').reset_index()

    return data_filtered
```

```python
# пока что запишем новый сет в отдельную переменную, чтобы можно было провести проверку корректности работы функций
new_data = aggregate(data)
```

Скорей всего это одни и те же игры, но просто в разных источниках жанр игры определялся по-разному.

Если у игры стоит жанр Misс (Miscellaneous - прочее, разное) и есть дубликат с альтернативным жанром, то альтернативный жанр можно считать более точным и заменить Misc на такой жанр.

Жанр Action также является более общим (включает в себя) по отношению к Adventure, Shooter или Platformer, однако более точным по сравнению с Misc.

При обработке неявных дубликатов, отдадим жанру Misc наименьший приоритет, жанру Adventure чуть более высокий приоритет чем Misc, но меньший чем другим жанрам.

```python
# определим функцию для замены жанров
def choose_best_genre(data):
    priority = {
        'Misc': 1,  # Приоритет для жанра "Misc" (Разное) равен 1
        'Action': 2  # Приоритет для жанра "Action" (Экшн) равен 2
    }
    default_priority = 3  # Приоритет по умолчанию для других жанров равен 3
    
    # Создаем копию данных, содержащих дубликаты (name, year_of_release, platform)
    duplicated_data = data[data.duplicated(subset=['name', 'year_of_release', 'platform'], keep=False)].copy()
    
    # Добавляем столбец 'priority', где значение определяется исходя из жанра игры и словаря 'priority'
    duplicated_data['priority'] = duplicated_data['genre'].apply(lambda x: priority.get(x, default_priority))
    
    # Сгруппировываем данные по названию игры, платформе и году выпуска, выбираем максимальный приоритет и соответствующий жанр
    best_genre = duplicated_data.groupby(['name', 'platform', 'year_of_release'])[['priority', 'genre']].max().reset_index()
    
    # Объединяем исходные данные с данными о лучшем жанре по выбранным столбцам, сохраняя все строки из исходных данных
    data = data.merge(best_genre, on=['name', 'platform', 'year_of_release'], how='left', suffixes=('_x', '_y'))
    
    # Заполняем пропущенные значения жанра из исходного столбца 'genre_x' в столбце 'genre_y'
    data['genre_y'] = data['genre_y'].fillna(data['genre_x'])
    
    # Заменяем исходный столбец 'genre_x' на столбец 'genre_y'
    data['genre_x'] = data['genre_y']
    
    # Удаляем столбцы 'genre_y' и 'priority' из данных
    data.drop(['genre_y', 'priority'], axis=1, inplace=True)
    
    # Переименовываем столбец 'genre_x' в 'genre'
    data.rename(columns={'genre_x': 'genre'}, inplace=True)
    
    return data
```

```python
# заменим жанры созданной функцией и сохраним обновленный датасет в новую переменную data2 (временно для целей дальнейшей проверки)
data2 = choose_best_genre(data)
```

```python
def plot_piechart(data, top=6, title='None'):
    # Проверка, если top меньше или равен количеству категорий
    if top >= len(data):
        # Построение круговой диаграммы без изменений
        plt.figure(figsize=(8, 6))
        plt.title(title)
        plt.pie(data, labels=data.index, autopct='%1.1f%%', startangle=90, counterclock=False)
        plt.axis('equal')
        plt.show()
    else:
        # Выделение топ N платформ и объединение остальных в категорию "Другие"
        top_platforms = data.tail(top)
        other_sales = data.head(len(data) - top).sum()
        top_platforms['Другие'] = other_sales

        # Построение круговой диаграммы
        plt.figure(figsize=(8, 6))
        plt.title(title)
        plt.pie(top_platforms, labels=top_platforms.index, autopct='%1.1f%%', startangle=90, counterclock=False)
        plt.axis('equal')
        plt.show()
```

```python
# Доля платформ из ТОП6 в общих продажах
plot_piechart(
    data.groupby('platform')['total_sales'].sum().sort_values(),
    top=6,
    title='Доля Топ-6 платформ в общих продажах')
```

Чтобы точней оценить срок жизни платформ, рассчитаем жизненный цикл платформы как с учетом выбросов, так и без.

```python
# функция для нахождения границы диаграммы размаха (границы выбросов)
def wiskers(column, df=data):
    # Расчет квартилей
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    # Расчет интерквартильного размаха (IQR)
    IQR = Q3 - Q1

    # Расчет границ ящика с усами
    lower_whisker = df[column][df[column] >= (Q1 - 1.5 * IQR)].min()
    upper_whisker = df[column][df[column] <= (Q3 + 1.5 * IQR)].max()

    return lower_whisker, upper_whisker
```

```python
# Расчет границ ящика с усами по каждой платформе и запись значений в соответствующие колонки таблицы
platform_lifecycle['lower_wisker'] = platform_lifecycle.apply(lambda row: wiskers('year_of_release', data[data['platform'] == row['platform']]), axis=1).str[0]
platform_lifecycle['upper_wisker'] = platform_lifecycle.apply(lambda row: wiskers('year_of_release', data[data['platform'] == row['platform']]), axis=1).str[1]
```

```python
# расчет жизненного цикла игровой платформы (полный от мин до макс) и без учета выбросов
platform_lifecycle['full_lifecycle'] = (platform_lifecycle['max'] - platform_lifecycle['min']).astype('int')
platform_lifecycle['wiskers_lifecycle'] = platform_lifecycle['upper_wisker'] - platform_lifecycle['lower_wisker']
```

```python
platform_lifecycle.head()
```

```python
platform_lifecycle[['full_lifecycle', 'wiskers_lifecycle']].describe()
```

```python
# построим график длительности жизненных циклов по платформам
plt.figure(figsize=(15, 6))


sns.lineplot(x='platform', y='full_lifecycle', data=platform_lifecycle, marker='o', label='Полный жизненный цикл (от мин до макс)')
sns.lineplot(x='platform', y='wiskers_lifecycle', data=platform_lifecycle, marker='o', label='Жизненный цикл без учета выбросов')


plt.xlabel('Платформа')
plt.ylabel('Жизненный цикл, лет')
plt.title('Жизненный цикл игровой платформы')


plt.legend()
plt.show()
```

Срок жизни игровых платформ в выборке не превышает 31 год (16 лет, если не учитывать выбросы).

В среднем жизненный цикл игровой платформы составляет около 6-7 лет.

Учитывая короткий жизненный цикл игровых платформ (6-7 лет для всех и около 10 лет для топовых платформ), а также частую смену поколений, исторический период, за который берутся данные для выбора перспективных платформ для 2017 год не должен превышать срока жизни последних поколений приставок.

#### Оценки пользователей и критиков

Посмотрим как зависят ли продажи игры от оценок пользователей и критиков. 
Для примера возьмем две самые современные и наиболее перспективные платформы PS4 и XOne.

```python
def scatterplot_and_correlation(platform, data=data):
    platform_scores = data.query('(year_of_release >= 2012) and (platform == @platform)')
    
    # Создание матричного графика рассеяния
    sns.set(style="ticks")
    sns.pairplot(platform_scores, vars=["total_sales", "critic_score", "user_score"])
    
    # Отображение графика
    plt.show()
    
    correlation = platform_scores[['critic_score', 'user_score', 'total_sales']].corr()
    return correlation # Отображение таблицы корреляции
```

```python
# данные по PS4
scatterplot_and_correlation('PS4')
```

```python
# Данные по XOne
scatterplot_and_correlation('XOne')
```

На матрицах рассеяния и в таблице корреляции видны следующие зависимости:
- Наблюдается умеренная положительная связь между оценкой критиков (critic_score) и общим количеством продаж игр на платформе (total_sales)
- Оценки пользователей практически (user_score) не влияют на общее количество продаж игр (total_sales).

## Проверка гипотез

### Средние пользовательские рейтинги платформ Xbox One и PC одинаковые

Проверим равны ли пользовательские рейтинги игр на платформах Xbox One и PC.

- H0 (нулевая гипотеза) = средний пользовательский рейтинг игр на платформах Xbox One и PC одинаковый.
- H1 (альтернативная гипотеза) = средний пользовательский рейтинг игр на платформах Xbox One и PC различается.

Оценки игр на Xbox One и PC не зависят друг от друга, поэтому для проверки гипотез можем применить двухвыборочный t-тест для двух независимых выборок.

```python
# создадим соответствующие выборки (удалим значения Nan и преобразуем в float)
xone_scores = actual_data.query('platform == "XOne"')['user_score'].dropna().astype('float')
pc_scores = actual_data.query('platform == "PC"')['user_score'].dropna().astype('float')
```

```python
#Проверим действительно ли выборки независимые
# Вычисление коэффициента корреляции с помощью метода corr()

print(f'Коэффициент корреляции: {(xone_scores.corr(pc_scores)):.2f}')
```

Корреляция отсутствует, выборки независимые.

```python
print(f'Средний пользовательский рейтинг игр на платформе Xbox One:', round(xone_scores.mean(), 2))
print(f'Средний пользовательский рейтинг игр на платформе PC:', round(pc_scores.mean(), 2))
```

```python
# выбранный уровень статистической значимости
alpha = 0.05
print(f"Уровень статистической значимости: {alpha * 100}%")

# проведем двусторонний ttest
results = st.ttest_ind(xone_scores, pc_scores)

#выведем полученное значение p-value
print('p-value:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу')
```

```python
# выбранный уровень статистической значимости
alpha = 0.01
print(f"Уровень статистической значимости: {alpha * 100}%")

# проведем двусторонний ttest
results = st.ttest_ind(xone_scores, pc_scores)

#выведем полученное значение p-value
print('p-value:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу')
```

<u>Результат проверки:</u> На имеющихся данных, при уровне значимости 1% (уровне доверия 99%) нет оснований отвергнуть нулевую гипотезу (H0) о равенстве средних пользовательских рейтингов игр на платформах Xbox One и PC. Соответственно, есть основания отвергнуть альтернативную гипотезу (H1) о том, что средние пользовательские рейтинги на данных платформах не равны.

### Средние пользовательские рейтинги жанров Action (англ. «действие», экшен-игры) и Sports (англ. «спортивные соревнования») разные.

Проверим равны ли пользовательские рейтинги игр жанров Action и Sports.

- H1 (альтернативная гипотеза) = средний пользовательский рейтинг игр жанра Action и Sports различается.

Оценки пользователей Action и Sports не зависят друг от друга, поэтому для проверки гипотез можем применить двухвыборочный t-тест для двух независимых выборок.

```python
# создадим соответствующие выборки (удалим значения Nan и преобразуем в float)
action_scores = actual_data.query('genre == "Action"')['user_score'].dropna().astype('float')
sports_scores = actual_data.query('genre == "Sports"')['user_score'].dropna().astype('float')
```

```python
#Проверим действительно ли выборки независимые
# Вычисление коэффициента корреляции с помощью метода corr()

print(f'Коэффициент корреляции: {(xone_scores.corr(pc_scores)):.2f}')
```

Корреляция отсутствует, выборки независимые.

```python
print(f'Средний пользовательский рейтинг игр жанра Action:', round(action_scores.mean(), 2))
print(f'Средний пользовательский рейтинг игр жанра Sports:', round(sports_scores.mean(), 2))
```

```python
# выбранный уровень статистической значимости
alpha = 0.05
print(f"Уровень статистической значимости: {alpha * 100}%")

# проведем двусторонний ttest
results = st.ttest_ind(action_scores, sports_scores)

#выведем полученное значение p-value
print('p-value:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу')
```

```python
# выбранный уровень статистической значимости
alpha = 0.01
print(f"Уровень статистической значимости: {alpha * 100}%")

# проведем двусторонний ttest
results = st.ttest_ind(action_scores, sports_scores)

#выведем полученное значение p-value
print('p-value:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу')
```

<u>Результаты проверки:</u> На имеющихся данных, при уровне значимости 1% (уровне доверия 99%) есть основания отвергнуть нулевую гиппотезу (H0) о равенстве средних пользовательских оценок в играх жанров Action и Sports. Соответственно, нет оснований отвергать альтернативную гипотезу (H1) о различии средних пользовательских оценок у игр жанров Action и Sports.
